BAR PLOT

```python
colors = a["species"].map({
    "Adelie": "blue",
    "Chinstrap": "green",
    "Gentoo": "red"
})
plt.figure(figsize=(6,5))
sns.barplot(x="species",y="body_mass_g",data=a)
plt.title("Bar Plot of Penguins")
plt.xlabel("Species")
plt.ylabel("Body Mass")
plt.show()
```

SCATTER PLOT

```python
colors = a["species"].map({
    "Adelie": "blue",
    "Chinstrap": "green",
    "Gentoo": "red"
})
plt.figure(figsize=(7,5))
plt.scatter(a["bill_length_mm"],a["bill_depth_mm"],c=colors)
plt.title("Bill Length vs Bill Depth")
plt.xlabel("Bill Length")
plt.ylabel("Bill Depth")
plt.show()
```

```python

fig = px.scatter(
    a,
    x="bill_length_mm",
    y="bill_depth_mm",
    color="species",
    color_discrete_map={
        "Adelie": "blue",
        "Chinstrap": "green",
        "Gentoo": "red"
    },
    title="Bill Length vs Bill Depth",
    labels={
        "bill_length_mm": "Bill Length",
        "bill_depth_mm": "Bill Depth",
        "species": "Species"
    }
)

fig.update_layout(
    width=700,
    height=500
)

fig.show()
```

PIE CHART

```python
s_c=a["species"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(
    s_c,labels=s_c.index,autopct="%1.1f%%",startangle=90)
plt.title("Distribution of Penguin Species")
plt.show()
```

HISTOGRAM

```python
plt.figure(figsize=(7,5))
plt.hist(a["flipper_length_mm"],bins=15,color="lightblue",edgecolor="black")
plt.title("Distribution of Flipper Length")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Frequency")
plt.show()
```

BOX PLOT

```python
plt.figure(figsize=(6,5))
sns.boxplot(x="species",y="body_mass_g",data=a)
plt.title("Body Mass Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Body Mass (g)")
plt.show()
```
